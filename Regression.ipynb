2. What are the key assumptions of Simple Linear Regression?
-> i. Linearity: The relationship between X and Y is linear.

ii. Independence: Observations are independent of each other.

iii. Homoscedasticity: Constant variance of errors (residuals) across all levels of X.

iv. Normality of Errors: The residuals (errors) are normally distributed.

v. No Multicollinearity: Not applicable here (since there's only one predictor in simple linear regression).

3. What does the coefficient m represent in the equation Y=mX+c?
-> In the equation Y = mX + c:

m is the slope of the line.

It represents the rate of change of the dependent variable Y with respect to the independent variable X.

In other words:
For every one-unit increase in X, Y changes by m units.

For example, if m = 2, then for every 1 unit increase in X, Y increases by 2 units.

4. What does the intercept c represent in the equation Y=mX+c?
-> In the equation Y = mX + c, the intercept c represents the value of Y when X is 0.

It shows where the line crosses the Y-axis.
For example, if c = 5, then when X = 0, Y = 5.

6. What is the purpose of the least squares method in Simple Linear Regression?
-> The purpose of the least squares method in Simple Linear Regression is to find the best-fitting line by minimizing the sum of the squared differences (errors) between the actual data points and the predicted values on the line.

In short: It makes the total prediction error as small as possible.

7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression?
-> In Simple Linear Regression, the coefficient of determination (R²) measures how well the regression line fits the data.

Interpretation (in short):
R² ranges from 0 to 1.

R² = 1 → perfect fit (100% of the variation in Y is explained by X).

R² = 0 → the model explains none of the variation in Y.

Higher R² means a better fit.

For example, R² = 0.85 means 85% of the variation in Y is explained by X.



8. **Multiple Linear Regression (MLR)**

**Definition**: MLR models the relationship between one dependent variable (Y) and **two or more independent variables (X₁, X₂, ..., Xₙ)**.

---

### **9. Main Difference: Simple vs. Multiple Linear Regression**

* **Simple**: One independent variable
* **Multiple**: Two or more independent variables

---

### **10. Key Assumptions of Multiple Linear Regression**

1. Linearity
2. Independence of errors
3. Homoscedasticity
4. Normality of residuals
5. No multicollinearity
6. No autocorrelation (especially in time series)

---

### **11. Heteroscedasticity and Its Effect**

**Heteroscedasticity** = non-constant variance of residuals.
**Effect**: Leads to **inefficient estimates** and **unreliable standard errors**, which can mislead hypothesis tests.

---

### **12. Improving High Multicollinearity**

* Remove or combine correlated predictors
* Use **Principal Component Analysis (PCA)**
* Apply **Ridge or Lasso regression**
* Center or standardize variables

---

### **13. Transforming Categorical Variables**

* **One-hot encoding** (most common)
* **Label encoding** (for ordinal data)
* **Binary encoding** or **target encoding** (for high-cardinality)

---

### **14. Role of Interaction Terms**

Interaction terms allow the effect of one variable to **depend on another**.
Useful when the combined influence of variables isn’t simply additive.

---

### **15. Intercept Interpretation: Simple vs. Multiple**

* **Simple**: Value of Y when X = 0
* **Multiple**: Value of Y when **all X₁, X₂, ..., Xₙ = 0** (which may not always be meaningful)

---

### **16. Significance of the Slope**

The **slope** shows how much Y changes with a **1-unit change in an independent variable**, holding others constant.
It directly affects predictions and variable importance.

---

### **17. Role of Intercept in Regression Context**

The intercept gives the **baseline level of Y** when all predictors are 0. It provides a starting point for predictions and context, though not always interpretable.

---

### **18. Limitations of R² as a Sole Measure**

* Doesn’t indicate **causation**
* Can increase with **irrelevant predictors**
* Doesn’t measure **model complexity** or **predictive accuracy**

Use **Adjusted R², RMSE, or cross-validation** too.

---

### **19. Large Standard Error of Coefficient**

* Suggests the coefficient estimate is **unstable**
* Could be due to **multicollinearity** or **small sample size**
* May lead to **insignificant p-values**

---

### **20. Identifying Heteroscedasticity in Residual Plots**

* Look for a **funnel shape** in residuals vs. fitted values
* Important because it violates assumptions, impacting inference

---

### **21. High R² but Low Adjusted R²**

Indicates that **non-informative variables** are in the model.
**Adjusted R²** penalizes for adding irrelevant predictors.

---

### **22. Scaling Variables in MLR**

Important when:

* Using **regularization (Ridge/Lasso)**
* Variables have **different units**
* Improves **interpretability** and **numerical stability**

---

### **23. What is Polynomial Regression?**

A form of regression where the **relationship between X and Y is modeled as an nth-degree polynomial**.

---

### **24. Polynomial vs. Linear Regression**

* **Linear**: Straight line
* **Polynomial**: Curved line (quadratic, cubic, etc.)
  Polynomial regression captures **nonlinear relationships**.

---

### **25. When to Use Polynomial Regression**

* When **scatter plots show curvature**
* To model **nonlinear trends** in the data

---

### **26. General Equation of Polynomial Regression**

$$
Y = b_0 + b_1X + b_2X^2 + b_3X^3 + ... + b_nX^n
$$

---

### **27. Can Polynomial Regression Use Multiple Variables?**

Yes! You can have multiple predictors, and apply polynomial terms to one or more.

---

### **28. Limitations of Polynomial Regression**

* Can **overfit** with high degree
* Sensitive to **outliers**
* **Harder to interpret**

---

### **29. Evaluating Model Fit for Polynomial Degree**

* Use **cross-validation**, **AIC/BIC**, or **adjusted R²**
* Plot **validation error vs. degree** to find the best fit

---

### **30. Importance of Visualization in Polynomial Regression**

* Helps detect **overfitting or underfitting**
* Visualizes **how well the curve follows the data**

---
